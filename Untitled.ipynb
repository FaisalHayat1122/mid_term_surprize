```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('Ecommerce Dataset.xlsx')
```

```python
df
```

```python
df.isnull().any()
```

```python
df.dropna(inplace=True)
```

```python
df.dtypes
```

```python
quantity_ordered_by_month=df.groupby(['Month'],as_index=False)['qty_ordered'].sum().sort_values(by='qty_ordered',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sns.countplot(data=df,x='category_name_1')
```

***Most Ordered Product***

```python
sc=sns.barplot(data=quantity_ordered_by_month,x='Month',y='qty_ordered',hue='Month')
```

***Quantity ordered refunded by month***

```python
sns.countplot(data=df,x='qty_ordered')
```

```python
new=df.groupby(['category_name_1'],as_index=False)['qty_ordered'].sum().sort_values(by='qty_ordered',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sc=sns.barplot(data=new,x='category_name_1',y='qty_ordered',hue='category_name_1')
```

***Top 10 ordered product***

```python
new=df.groupby(['category_name_1'],as_index=False)['qty_ordered'].sum().sort_values(by='qty_ordered',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sc=sns.barplot(data=new,x='category_name_1',y='qty_ordered',hue='category_name_1')
```

***Lowest 10 ordered Product***

```python
new=df.groupby(['category_name_1'],as_index=False)['qty_ordered'].sum().sort_values(by='qty_ordered',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sc=sns.barplot(data=new,x='category_name_1',y='qty_ordered',hue='category_name_1')
```

***Most Category Orders Sold***

```python
new=df.groupby(['category_name_1'],as_index=False)['price'].sum().sort_values(by='price',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sc=sns.barplot(data=new,x='category_name_1',y='price',hue='category_name_1')
```

***Category Wise Sales***

```python
new=df.groupby(['Month'],as_index=False)['price'].sum().sort_values(by='price',ascending=False)
```

```python
plt.figure(figsize=(30,5))
sc=sns.barplot(data=new,x='Month',y='price',hue='Month')
```
